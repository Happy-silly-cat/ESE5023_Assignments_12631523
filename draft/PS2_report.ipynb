## 2. Air temperature in Shenzhen, 1998–2022

TMP 字段由温度值和质量代码组成。温度值需要除以 10 才能转换为摄氏度；质量代码保留为 1 或 5 的记录；+9999 表示缺失温度，予以删除。原始数据共有 235,674 条小时记录，清理后保留 234,864 条有效记录，形成 298 个月的月平均温度，时间范围为 1998 年 1 月至 2022 年 10 月。

月平均温度最低约为 12.18 °C，出现在 2011 年 1 月；最高约为 30.33 °C，出现在 2020 年 7 月。温度有明显的季节变化，夏季高、冬季低。

从年度平均温度看，1998 年约为 24.07 °C，2022 年约为 24.48 °C，线性趋势约为每年升高 0.029 °C，总变化约 0.41 °C。因此，1998–2022 年期间存在轻微升温趋势，但季节变化明显强于长期变化。2022 年数据只到 10 月，因此该年的年度平均值需要谨慎解释。

### 3.7 每日数据点数量

### 3.8 日序气候态

按一年中的第几天计算每日数据点数量的平均值，得到 366 个日序气候态值。数据点数量在一年中的分布不均匀，通常夏季较多、冬季较少。当前数据中，平均数据点数量最高约出现在第 222 天，最低约出现在第 37 天。

### 3.9 每日异常值

每日异常值等于当天数据点数量减去该日序对应的气候态平均值。正异常表示当天的观测点多于该日期的长期平均水平，负异常表示少于长期平均水平。这个指标能够去除季节循环后，比较不同年份和不同日期的异常活动。

## 4. Explore a data set: historical tsunami records

本题使用海啸历史数据文件进行探索。文件为制表符分隔的数据，包含时间、国家、地点、经纬度、地震震级、最大水高和死亡人数等字段。

### 4.1 数据读取和清理

数据共有 3,138 条记录和 52 个字段，年份范围为公元前 2000 年至 2026 年。清理时跳过下载条件说明行，将 Year、Earthquake Magnitude、Maximum Water Height 和 Total Deaths 转换为数值，并删除缺失年份记录。数值字段中的缺失值保留为缺失，在进行相应统计时排除。

### 4.2 时间序列

按年份统计海啸记录数量后，共有 585 个有记录的年份。记录数最多的年份是 1868、1928 和 1938 年，每年 26 条。年度记录数量分布不均匀，近现代记录通常更完整，因此时间序列的变化同时受到真实海啸活动和历史观测、数据保存程度的影响。不能仅凭记录数量增加，就断定海啸发生频率一定增加。

### 4.3 统计检查

对年度海啸记录数进行了 8 项统计检查：

| 统计量 | 结果 |
|---|---:|
| 有记录的年份数 | 585 |
| 平均每年记录数 | 5.36 |
| 中位数 | 2 |
| 标准差 | 5.90 |
| 最小值 | 1 |
| 最大值 | 26 |
| 第一四分位数 | 1 |
| 第三四分位数 | 8 |

平均值高于中位数，说明年度记录数分布右偏，少数记录较多的年份明显抬高了平均值。

### 补充：震级与死亡人数

在震级和死亡人数同时有记录的 558 个事件中，Pearson 相关系数为 0.108，Spearman 相关系数为 0.213，说明两者只有弱的正相关。只统计 1900 年以后时，369 个事件的 Pearson 相关系数为 0.132，Spearman 相关系数为 0.260，结论仍然是弱正相关。

死亡人数还受到人口密度、受灾地点、海啸高度、预警系统和建筑抗震能力等因素影响，因此震级不能单独决定死亡人数。缺失死亡人数没有被当作零，而是在相关性计算中排除。

### 补充：震级与最大水高

在 1,011 个震级和最大水高同时有记录的事件中，Pearson 相关系数为 0.138，Spearman 相关系数为 0.360，说明存在弱到中等程度的正向单调关系，但线性关系较弱。只统计 1900 年以后时，Pearson 相关系数为 0.125，Spearman 相关系数为 0.335。

最大水高还受到海啸产生机制、震源距离、海底地形、海岸形状和观测位置等因素影响，因此不能只根据震级预测最大水高。

### 补充：海啸发生地点地图

数据中有 2,672 条记录同时包含纬度和经度，已在原 PS2.ipynb 中绘制世界地图。部分点位于内陆并不一定是绘图错误，因为数据中也包含湖泊、河流、火山或滑坡等非海洋水体事件，历史记录的坐标也可能是根据历史描述推定的近似位置。若只想分析海洋海啸，应使用 Oceanic Tsunami 字段进行筛选。
